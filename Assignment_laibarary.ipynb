```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("//content/data.csv")
df
```

```python
df.info()
```

```python
df.describe()
```

```python

df.isnull().sum()
```

```python
df['Engine HP'].fillna(df["Engine HP"].median(),inplace=True)
```

```python

df.isnull().sum()
```

```python
df['Engine Cylinders'].fillna(df['Engine Cylinders'].median(),inplace=True)
```

```python

df.isnull().sum()
```

```python
df['Engine Fuel Type'].fillna(df['Engine Fuel Type'].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Number of Doors'].fillna(df["Number of Doors"].mean(),inplace=True)
```

```python

df.isnull().sum()
```

```python
df['Market Category'].fillna(df['Market Category'].mode()[0],inplace=True)
```

```python

df.isnull().sum()
```

```python

df.info()
```

```python

df.drop_duplicates()
```

```python
df['Vehicle Size'].value_counts().plot.pie(autopct='%1.1f%%', figsize=(6,6))
plt.title('Distribution of Vehicle Size')
plt.ylabel('')
plt.show()
```

```python

df3=df.groupby("Make")["Number of Doors"].sum()
df3

```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Year', y='Popularity', hue='Transmission Type')
plt.title('Popularity of Cars by Year and Transmission Type', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Popularity')
plt.xticks(rotation=45)
plt.legend(title='Transmission Type')
plt.tight_layout()
plt.show()
```

```python
sns.displot(data=df, x ='Year', kind='hist',hue='Transmission Type')
```

```python
# top 10 expensive
a= df.groupby("Make")["MSRP"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=a.values, y=a.index, palette="viridis")

plt.title("Top 10 Car Brands by Average Price (MSRP)")
plt.xlabel("Average MSRP ($)")
plt.ylabel("Car Make")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10,6))
sns.boxplot(data=df, x="Vehicle Size", y="MSRP", palette="Set2")

plt.title("Price Distribution by Vehicle Size")
plt.ylabel("MSRP ($)")
plt.xlabel("Vehicle Size")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
b = df.groupby("Vehicle Size")["city mpg"].mean().sort_values()

sns.barplot(x=b.index, y=b.values, palette="coolwarm")

plt.title("Average City MPG by Vehicle Size")
plt.ylabel("Average City MPG")
plt.xlabel("Vehicle Size")
plt.tight_layout()
plt.show()
```

```python

n= df.groupby("Make")["Engine HP"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=n.values, y=n.index, palette="magma")

plt.title("Top 10 Car by Average Engine ")
plt.xlabel("Average Engine Horsepower")
plt.ylabel("Car Make")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df["highway MPG"], bins=20, color="teal", kde=True)

plt.title("Distribution of Highway MPG")
plt.xlabel("Highway MPG")
plt.ylabel("Number of Cars")
plt.tight_layout()
plt.show()
```
