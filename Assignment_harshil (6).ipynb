80)Write a Python program to count the frequency of words in a file.

```python
def word(file_name):
    with open(file_name, 'r') as file:
        text = file.read()
        words = text.split()
        freq = {}

        for word in words:
            if word in freq:
                freq[word] += 1
            else:
                freq[word] = 1
    return freq

file_name = 'harshil.txt'
count = word(file_name)
print(count)
```

89)How Do You Handle Exceptions with Try/Except/Finally in Python?
Explain with coding snippets.

Ans:-

try = here we write code which give error.

except: If any error come then its use to handle.

finally: This block always execute if there is error or not.


```python


try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result:", result)
except ZeroDivisionError:
    print("0 its not divisible")
except ValueError:
    print("write only number")
finally:
    print("Try-Except-Block is over")
```

90)Write python program that user to enter only odd numbers, else will raise an exception.

```python
try:
    n = int(input("Enter an odd number: "))

    if n % 2 == 0:
        raise Exception("Even number not allowed")

    print("Valid odd number:", n)

except ValueError:
    print("write only number not text!")

except Exception as e:
    print("exception:", e)

finally:
    print("Program is end")
```
